```python
import torch
from torch import nn
from torch.nn import functional as F

x = torch.arange(4)
torch.save(x, 'x-file')
```

```python
x2 = torch.load('x-file')
x2
```

- torch.save()：将张量序列化到磁盘文件
- torch.load()：从磁盘反序列化恢复张量

```python
y = torch.zeros(4)
torch.save([x, y],'x-files')
x2, y2 = torch.load('x-files')
(x2, y2)
```

可以直接保存列表、元组等容器，加载后会按原结构返回

```python
mydict = {'x': x, 'y': y}
torch.save(mydict, 'mydict')
mydict2 = torch.load('mydict')
mydict2
```

用字典保存可以为张量命名，加载后通过键值对访问，可读性更强

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(20, 256)
        self.output = nn.Linear(256, 10)

    def forward(self, x):
        return self.output(F.relu(self.hidden(x)))

net = MLP()
X = torch.randn(size=(2, 20))
Y = net(X)
```

定义了一个简单的多层感知机（MLP），包含 1 个隐藏层和 1 个输出层
前向传播使用 ReLU 激活函数，将 20 维输入映射到 10 维输出

```python
torch.save(net.state_dict(), 'mlp.params')
```

```python
clone = MLP()
clone.load_state_dict(torch.load('mlp.params'))
clone.eval()
```

```python
Y_clone = clone(X)
Y_clone == Y
```
